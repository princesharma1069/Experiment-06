```python
''' 7) Problem  statement:
write a Python program to calculate the gross salary of an employee based
if basicc <= 10000: HRA = 20%, DA = 80%
if basicc <= 20000: HRA = 25%, DA = 90%
otherwise: HRA = 30%, DA = 95%'''
salary = int(input("Enter basic salary(in Rs):"))
if salary <=10000:
  salary = salary +((20/100)*salary)+((80/100)*salary)
  print("Gross salary(in Rs):",salary)
elif salary <=20000:
  salary = salary +((25/100)*salary)+((90/100)*salary)
  print("Gross salary(in Rs):",salary)
else:
  salary = salary +((30/100)*salary)+((95/100)*salary)
  print("Gross salary(in Rs):",salary)
```

```python
'''8) write a python program to calculate income tax based on annual income
upto Rs 2,50,000: No tax
Rs 2,50,001- Rs 5,00,000: 5%
Rs 5,00,001- Rs 10,00,000: 20%
Above Rs 10,00,000: 30% '''
Income = int(input("Enter Income(in Rs):"))
if Income <= 250000:
  print("Income tax= 0%")
elif Income >= 250001 and Income <= 500000:
  tax = (Income -250000) * (5/100)
  print("Income tax(in Rs):",tax)
elif Income >= 500001 and Income <= 1000000:
  tax = (500000 -250000) * (5/100) + (Income - 500000) * (20/100)
  print("Income tax(in Rs):",tax)
else:
  tax = (500000 -250000) * (5/100) + (1000000 - 500000) * (20/100) + (Income - 1000000) * (30/100)
  print("Income tax(in Rs):",tax)
```
